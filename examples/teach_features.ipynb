# Feature Engineering Toolkit — Beginner Guide

Teaches basic Python users how to create, select, and transform features for ML.

**Prerequisite:** basic Python + pandas. Install with `pip install -e .`.

```python
import pandas as pd
import numpy as np
from feature_engineering_toolkit import FeatureEngineeringPipeline

print('Imports OK!')
```

## Step 1 — Make sample data

We create a small dataset with a target column (0/1) and features we can engineer.

```python
np.random.seed(42)
df = pd.DataFrame({
    'user_id': list(range(1, 21)) * 3,
    'age': list(range(20, 40)) * 3,
    'income': [50000 + i*1000 for i in range(60)],
    'date': pd.date_range('2024-01-01', periods=60, freq='D'),
    'target': [0, 1] * 30,
})
print('Shape:', df.shape)
df.head()
```

## Step 2 — Run the pipeline

The pipeline does everything: create → transform → select → evaluate.

Order matters: **create → transform → select → evaluate**.

```python
pipeline = FeatureEngineeringPipeline(df, target_col='target')
pipeline.run()

print('\n=== REPORT ===')
print(pipeline.generate_report())
```

## Step 3 — Inspect what was created

The pipeline logs every step. Check the created/selected features:

```python
print('Created features:', pipeline.created_features)
print('Selected features:', pipeline.selected_features)
print('Model accuracy:', pipeline.accuracy)
print('\nEngineered data shape:', pipeline.get_engineered_data().shape)
```

## Step 4 — Save the result

Save the engineered dataset and report:

```python
pipeline.save_data('engineered_output.csv')
print('Saved engineered_output.csv')
```
