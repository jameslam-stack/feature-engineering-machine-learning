# Engineer a CSV File — File Workflow

Full workflow: load messy CSV → engineer features → evaluate → save.

**Prerequisite:** Run `teach_features.ipynb` first.

```python
import pandas as pd
from feature_engineering_toolkit import FeatureEngineeringPipeline

# Load the messy CSV
messy = pd.read_csv('data/input.csv')
print('Loaded:', messy.shape)
messy.head()
```

```python
# Inspect problems
print('Missing:\n', messy.isnull().sum())
print('\nOut-of-range age:', (messy.age > 120).sum())
print('Negative income:', (messy.income < 0).sum())
```

```python
# Run the full pipeline
pipeline = FeatureEngineeringPipeline(messy, target_col='target')
pipeline.run()

print('\n=== REPORT ===')
print(pipeline.generate_report())
```

```python
# Inspect what was created
print('Created:', pipeline.created_features)
print('Selected:', pipeline.selected_features)
print('Accuracy:', pipeline.accuracy)
print('\nEngineered shape:', pipeline.get_engineered_data().shape)
```

```python
# Save the result
pipeline.save_data('data/processed.csv')
print('Saved data/processed.csv')
```

```python
# Verify by reloading
reloaded = pd.read_csv('data/processed.csv')
print('Reloaded:', reloaded.shape)
reloaded.head()
```

## Summary

| Step | Method | What it did |
|---|---|---|
| 1 | `pd.read_csv()` | Load file |
| 2 | `.create_features()` | Polynomial, interaction, aggregation, datetime |
| 3 | `.transform()` | Scaling (StandardScaler) |
| 4 | `.select_features()` | SelectKBest top features |
| 5 | `.evaluate()` | Model accuracy |
| 6 | `.save_data()` | Write result |

**Tip:** Edit `config_template.yaml` to change methods (e.g., `robust_scaler`, `recursive_feature_elimination`).
